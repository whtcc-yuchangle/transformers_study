```python
from transformers import pipeline
```

```python
pipe = pipeline(task="automatic-speech-recognition", model="openai/whisper-large-v3")
```

1. 下载ffmpeg       https://ffmpeg.org/download.html#build-windows      ffmpeg-master-latest-win64-gpl.zip
2. 解压到固定路径（如 D:\ffmpeg）
3. 将 ffmpeg\bin 路径（如 D:\ffmpeg\bin）添加到系统环境变量 Path
